```python
print("24BAD010-Anusri T")
import nltk
from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.probability import FreqDist
import string

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

sentences = brown.sents(categories='news')
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

preprocessed_sentences = []

for sentence in sentences:
    text = ' '.join(sentence).lower()
    tokens = word_tokenize(text)
    tokens = [
        word for word in tokens
        if word not in string.punctuation
        and not all(char in string.punctuation for char in word)
        and word not in stop_words
    ]
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    if tokens:
        preprocessed_sentences.append(tokens)

all_tokens = [
    word
    for sentence in preprocessed_sentences
    for word in sentence
]

vocabulary = set(all_tokens)
freq_dist = FreqDist(all_tokens)

print("Vocabulary Size:", len(vocabulary))

print("\nTop 10 Most Frequent Words:")
for word, frequency in freq_dist.most_common(10):
    print(word, ":", frequency)

print("\nFrequency Distribution:")
print(freq_dist)

print("\nTokenized Sentences for Word2Vec Training:")
for sentence in preprocessed_sentences[:5]:
    print(sentence)
```

```python
print("24BAD010-Anusri T")
import nltk
import string
from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

sentences = brown.sents(categories='news')
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
preprocessed_sentences = []

for sentence in sentences:
    text = ' '.join(sentence).lower()
    tokens = word_tokenize(text)
    tokens = [
        word for word in tokens
        if word not in string.punctuation
        and not all(char in string.punctuation for char in word)
        and word not in stop_words
    ]
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    if tokens:
        preprocessed_sentences.append(tokens)

model = Word2Vec(
    sentences=preprocessed_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=1,
    workers=4,
    seed=42,
    epochs=10
)

word_pairs = [
    ('government', 'president'),
    ('city', 'state'),
    ('school', 'student')
]

similarity_scores = {}

for word1, word2 in word_pairs:
    if word1 in model.wv and word2 in model.wv:
        vector1 = model.wv[word1].reshape(1, -1)
        vector2 = model.wv[word2].reshape(1, -1)

        score = cosine_similarity(vector1, vector2)[0][0]
        similarity_scores[(word1, word2)] = score

        print(f"Cosine Similarity between '{word1}' and '{word2}': {score:.4f}")
    else:
        print(f"Word pair '{word1}', '{word2}' not found in vocabulary.")

if similarity_scores:
    most_similar_pair = max(similarity_scores, key=similarity_scores.get)

    print("\nMost Similar Word Pair:", most_similar_pair)
    print("Highest Similarity Score:", round(similarity_scores[most_similar_pair], 4))

    print("\nSimilarity Comparison:")
    for pair, score in sorted(
        similarity_scores.items(),
        key=lambda item: item[1],
        reverse=True
    ):
        print(pair, ":", round(score, 4))

    print("\nObservations:")
    print("Higher cosine similarity indicates more similar word contexts.")
    print("Word2Vec learns relationships based on words appearing in similar contexts.")
    print("The pair with the highest score is the most similar among the selected pairs.")
```
