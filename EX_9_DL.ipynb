```python
print("24BAD010-Anusri T")
!pip install transformers torch -q

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

prompts = [
    "What is Artificial Intelligence?",
    "Explain Machine Learning in simple words.",
    "What are the applications of Deep Learning?"
]

for prompt in prompts:
    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        temperature=0.7,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)

    print("User:", prompt)
    print("LLM Response:", response)
    print("-" * 60)

print("LLM loading and response generation completed successfully!")
```

```python
print("24BAD010-Anusri T")
conversation_history = []

def chatbot(user_query):
    global conversation_history

    prompt = ""
    for user, bot in conversation_history[-3:]:
        prompt += f"User: {user}\nBot: {bot}\n"

    prompt += f"User: {user_query}\nBot:"

    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=800)

    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        temperature=0.7,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id,
        repetition_penalty=1.2
    )

    response = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    response = response.split("User:")[0].strip()

    conversation_history.append((user_query, response))

    return response


print("Chatbot is ready! Type 'exit' to stop.")

while True:
    query = input("You: ")

    if query.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    answer = chatbot(query)
    print("Chatbot:", answer)
```
